```python
import re
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report
import numpy as np
```

```python
def preprocess_text(text):
    text = text.lower()
    text = re.sub(r'[^a-z0-9\s]', '', text)
    text = re.sub(r'\s+', ' ', text).strip()
    return text
```

```python
def preprocess_text(text):
    text = text.lower()
    text = re.sub(r'[^a-z0-9\s]', '', text)
    text = re.sub(r'\s+', ' ', text).strip()
    return text
```

```python
def build_naive_bayes_classifier(positive_data, negative_data):

    processed_positive = [preprocess_text(review) for review in positive_data]
    processed_negative = [preprocess_text(review) for review in negative_data]
    print("Processed Positive Reviews:", processed_positive)
    print("Processed Negative Reviews:", processed_negative)
    
    
    all_reviews = processed_positive + processed_negative
    labels = np.array([1] * len(processed_positive) + [0] * len(processed_negative))

    vectorizer = CountVectorizer()
    X_counts = vectorizer.fit_transform(all_reviews)

    classifier = MultinomialNB()
    classifier.fit(X_counts, labels)

    return classifier, vectorizer

```

```python
def predict_sentiments(classifier, vectorizer, test_data):

    processed_test_data = [preprocess_text(review) for review in test_data]
    test_counts = vectorizer.transform(processed_test_data)
    
    predictions_numeric = classifier.predict(test_counts)
    
    sentiments = ["positive" if pred == 1 else "negative" for pred in predictions_numeric]
    return sentiments
```

```python
def evaluate_classifier(predictions, true_labels):

    if not isinstance(predictions, list): predictions = list(predictions)
    if not isinstance(true_labels, list): true_labels = list(true_labels)
        
    return accuracy_score(true_labels, predictions)
```

```python

#Classifier
classifier, vectorizer = build_naive_bayes_classifier(positive_reviews, negative_reviews)

predicted_sentiments = predict_sentiments(classifier, vectorizer, test_reviews)
print("Test reviews:")
for i, review in enumerate(test_reviews):
    print(f"'{review}' ! Predicted: {predicted_sentiments[i]}")

print("Evaluating:")
true_labels_test = ["positive", "negative", "negative", "negative"]

accuracy = evaluate_classifier(predicted_sentiments, true_labels_test)
print(f"Accuracy: {accuracy}")



```
